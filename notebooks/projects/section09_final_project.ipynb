This is not a prediction of whether the customer will ever cancel.

### Build a cutoff-safe customer table

Use `Customer ID` to assemble one row per eligible customer. Include members who signed up **before June 1** and had not cancelled before June 1. Build `y` as 1 for a cancellation on or after June 1 and before June 4, and 0 otherwise. This assumes cancellation records are complete through June 3.

Build exactly two predictors:

- `Listening Sessions`: join listening history to session login timestamps by `Session ID`. For each customer, count distinct session IDs with login timestamps on or after March 1 and before June 1, 2023. Use login time as a proxy for when the session started; multiple listening rows can belong to the same session. Keep eligible customers without qualifying sessions and give them a count of zero only if the logs are complete for the window.
